FCFS Disk Scheduling

```python
req_seq = [0, 13, 41, 50, 63, 89, 100]
head = 46
totalseek = 0
for i in req_seq:
  totalseek += abs( i-head )
  head = i
  print("Total head movement:",totalseek)
```

SSTF Disk Scheduling

```python
request = [0, 13, 41, 50, 63, 89, 100]
head = 46
total = 0

print("Seek Sequence:", head, end=" ->")

while request:
  nearest = min(request, key=lambda x: abs(x - head))
  total += abs(nearest - head)
  head = nearest
  request.remove(nearest)
  print(head, end=" ->")

print("\nTotal Seek Time:", total)
```

SCAN Scheduling (initially left)

```python
request = [0, 13, 41, 50, 63, 89, 100]
head = 46
total = 0

print("Seek Sequence:", head, end=" ->")

right = sorted([x for x in request if x >= head])
left = sorted([x for x in request if x < head], reverse=True)

for x in left:
  total += abs(x - head)
  head = x
  print(head, end=" ->")

for x in right:
  total += abs(x - head)
  head = x
  print(head, end=" ->")


print("\nTotal Seek Time:", total)
```

SCAN Scheduling (initially right)

```python
request = [0, 13, 41, 50, 63, 89, 100]
head = 46
total = 0

print("Seek Sequence:", head, end=" ->")

right = sorted([x for x in request if x >= head])
left = sorted([x for x in request if x < head], reverse=True)

for x in right:
  total += abs(x - head)
  head = x
  print(head, end=" ->")

for x in left:
  total += abs(x - head)
  head = x
  print(head, end=" ->")

print("\nTotal Seek Time:", total)
```
