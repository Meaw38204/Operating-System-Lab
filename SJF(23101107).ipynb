```python
processes = [
    ["P1", 3, 5],
    ["P2", 2, 2],
    ["P3", 3, 3],
    ["P4", 6, 4],
    ["P5", 2, 9],
    ["P6", 1, 5],
]

n = len(processes)
Time = 0
Completed = 0
Done = [False] * n
WaitingTime = [0] * n
TurnAroundTime = [0] * n
CompletionTime = [0] * n

while Completed < n:
  Available = []
  for i in range(n):
    if processes[i][1] <= Time and not Done[i]:
      Available.append(i)
  if not Available:
    Time+=1
    continue

  index = min(Available, key=lambda i:processes[i][2])
  p_id, AT, BT = processes[index]
  Time += BT
  CompletionTime[index] = Time
  TurnAroundTime[index] = CompletionTime[index] - AT
  WaitingTime[index] = TurnAroundTime[index] - BT

  Done[index] = True
  Completed += 1


print("processes\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
  print(f"{processes[i][0]}\t\t{processes[i][1]}\t{processes[i][2]}\t{CompletionTime[i]}\t{TurnAroundTime[i]}\t{WaitingTime[i]}")

print("Avg TAT =", sum(TurnAroundTime) / n)
print("Avg WT =", sum(WaitingTime) / n)
```

Non Preemptive SJF

```python
processes = [
    ["P1", 3, 5],
    ["P2", 2, 2],
    ["P3", 3, 3],
    ["P4", 6, 4],
    ["P5", 2, 9],
    ["P6", 1, 5],
]

n = len(processes)
RemainingTime = [p[2] for p in processes]
CompletionTime = [0] * n
TurnAroundTime = [0] * n
WaitingTime = [0] * n
Time = 0
Completed = 0

while Completed < n:
  Available = []
  for i in range(n):
    if processes[i][1] <= Time and RemainingTime[i] > 0:
      Available.append(i)

  if not Available:
    Time += 1
    continue


  index = min(Available, key=lambda i:(RemainingTime[i], processes[i][1]))
  RemainingTime[index] -= 1
  Time += 1

  if RemainingTime[index] == 0:
    CompletionTime[index] = Time
    TurnAroundTime[index] = CompletionTime[index] - processes[index][1]
    WaitingTime[index] = TurnAroundTime[index] - processes[index][2]
    Completed += 1


print("processes\tAT\tBT\tCT\tTAT\tWT")
for i in range(n):
  print(f"{processes[i][0]}\t\t{processes[i][1]}\t{processes[i][2]}\t{CompletionTime[i]}\t{TurnAroundTime[i]}\t{WaitingTime[i]}")

print("Avg TAT =", sum(TurnAroundTime) / n)
print("Avg WT =", sum(WaitingTime) / n)
```
