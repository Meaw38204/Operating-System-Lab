```python
class Process:
    def __init__(self, P_id, AT, BT):
        self.P_id = P_id
        self.AT = AT
        self.BT = BT
        self.CT = 0
        self.TAT = 0
        self.WT = 0

    def calculate(self):
        self.TAT = self.CT - self.AT
        self.WT = self.TAT - self.BT

    def get(self):
        print(f"P{self.P_id}\t{self.AT}\t{self.BT}\t{self.CT}\t{self.TAT}\t{self.WT}")

    def turnaround(self):
        return self.TAT

    def waiting(self):
        return self.WT


n = int(input("Enter the Number of Processes: "))
processes = []

# Input
for i in range(n):
    print(f"\nProcess {i+1}")
    AT = int(input("Enter Arrival Time: "))
    BT = int(input("Enter Burst Time: "))
    processes.append(Process(i + 1,AT, BT))

# Bubble Sort by Arrival Time
for i in range(n - 1):
    for j in range(n - i - 1):
        if processes[j].AT > processes[j + 1].AT:
            temp = processes[j]
            processes[j] = processes[j + 1]
            processes[j + 1] = temp

# Calculate Completion Time
current_time = 0

for p in processes:
    if current_time < p.AT:
        current_time = p.AT

    current_time += p.BT
    p.CT = current_time
    p.calculate()

# Output
print("\nP_id\tAT\tBT\tCT\tTAT\tWT")

AVG_TAT = 0
AVG_WT = 0

for p in processes:
    p.get()
    AVG_TAT += p.turnaround()
    AVG_WT += p.waiting()

print(f"\nAverage Turnaround Time = {AVG_TAT / n:.2f}")
print(f"Average Waiting Time = {AVG_WT / n:.2f}")
```
